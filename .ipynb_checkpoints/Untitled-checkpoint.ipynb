```python
import pandas as pd
import numpy as np

def gaussian_smoothing(df:pd.DataFrame, points:list, x:str = "x", y:str = "y", std:float = 3):
    """
    Suavização Gaussiana genérica (Nadaraya-Watson).
    :param x: Pode ser uma string 'x' (1D) ou uma lista de strings ['x1', 'x2', ...] (ND).
    :param points: Se 1D, lista de números. Se ND, lista de tuplas/listas com N coordenadas.
    """
    colunas_x = [x] if isinstance(x, str) else list(x)
    is_1d = len(colunas_x) == 1
    
    variancia = std * std
    limite = std * 3.5
    e = 2.718281828459045

    X_treino = df[colunas_x].values 
    Y_treino = df[y].values

    real = {col: [] for col in colunas_x}
    real.update({y: [], "std": []})

    for p in points:
        p_vetor = np.array([p]) if is_1d else np.array(p)

        dentro_do_limite = np.all(np.abs(X_treino - p_vetor) < limite, axis=1)
        
        X_vizinhos = X_treino[dentro_do_limite]
        Y_vizinhos = Y_treino[dentro_do_limite]

        if len(Y_vizinhos) == 0:
            for idx, col in enumerate(colunas_x):
                real[col].append(p_vetor[idx])
            real[y].append(0.0)
            real["std"].append(0.0)
            continue

        distancias_quadradas = np.sum((X_vizinhos - p_vetor) ** 2, axis=1)
        pesos = e ** (-1/2 * distancias_quadradas / variancia)
        sum_pesos = np.sum(pesos)

        if sum_pesos == 0:
            sum_pesos = 1e-10

        y_estimado = np.sum(Y_vizinhos * pesos) / sum_pesos
        std_estimado = np.sum(np.abs(y_estimado - Y_vizinhos) * pesos) / sum_pesos

        for idx, col in enumerate(colunas_x):
            real[col].append(p_vetor[idx])
        real[y].append(y_estimado)
        real["std"].append(std_estimado)

    return pd.DataFrame(real)


def kernel_smoothing(df:pd.DataFrame, points:list, x:str = "x", y:str = "y", std:float = 3, iterations:int = 20):
    """
    Ajuste recursivo (Gradient Boosting Residual) que se auto-adapta para 1D ou ND.
    """
    colunas_x = [x] if isinstance(x, str) else list(x)
    
    df_est_total = gaussian_smoothing(df, points=points, x=x, y=y, std=std)
    
    X_originais = df[colunas_x].values
    points_originais = list(X_originais[:, 0]) if len(colunas_x) == 1 else [tuple(row) for row in X_originais]
    
    df_pred_points_originais = gaussian_smoothing(df, points=points_originais, x=x, y=y, std=std)
    
    erro_puro = df[y].values - df_pred_points_originais[y].values
    dados_residuos = {col: df[col].values for col in colunas_x}
    dados_residuos[y] = erro_puro
    df_residuos = pd.DataFrame(dados_residuos)
    
    for _ in range(iterations - 1):
        df_est_erro = gaussian_smoothing(df_residuos, points=points, x=x, y=y, std=std)
        df_est_total[y] += df_est_erro[y].values
        
        df_pred_erro_originais = gaussian_smoothing(df_residuos, points=points_originais, x=x, y=y, std=std)
        df_residuos[y] = df_residuos[y].values - df_pred_erro_originais[y].values
        
    return df_est_total
```

```python
from math import cos
from random import random, seed
import pandas as pd
import matplotlib.pyplot as plt 

seed(1)
df = {"x":[*[i+random()*2-1 for i in range(100)], *[i+random()*2-1 for i in range(100)]],
      "y":[*[cos(i/12) + cos(i/5) + random()*2-1 for i in range(100)], *[cos(i/12) + cos(i/5) + random()*2-1 for i in range(100)]],}

df_real = pd.DataFrame({"x":[i for i in range(100)],
                       "y":[cos(i/12) + cos(i/5) for i in range(100)]})

df = pd.DataFrame(df)

resp = kernel_smoothing(df, points = [i/10 for i in range(100*10)], std = 6)

plt.figure(figsize = (9, 5))

plt.scatter(df["x"], df["y"], color = "gray", alpha = 0.4, label = "Samples")
plt.plot(df_real["x"], df_real["y"], color = "blue", linestyle = ":", alpha = 0.8, label = "Generating function")
plt.plot(resp["x"], resp["y"], color = "red", linestyle = "--", linewidth = 2, label = "Kernel regression estimation")
plt.fill_between(resp["x"], resp["y"] - resp["std"], resp["y"] + resp["std"], color = "green", alpha = 0.1, label = "Standard deviation")

plt.legend()
plt.ylabel("y")
plt.xlabel("x")
plt.show()
```

```python
from math import cos, sin
from random import random, seed
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

seed(1)
n_lado = 25  # Total de 625 points amostrados com ruído

# 1. Gerando a base de dados sintética com dois X (x1 e x2)
dados = {"x1": [], "x2": [], "y": []}
for i in range(n_lado):
    for j in range(n_lado):
        # Coordenadas com leve ruído espacial para simular amostras reais
        x1_val = i + random() * 0.4 - 0.2
        x2_val = j + random() * 0.4 - 0.2
        
        # Função geradora: Superfície matemática com ruído em Y
        y_real = cos(x1_val / 3) + sin(x2_val / 3)
        y_ruido = y_real + (random() * 0.6 - 0.3)
        
        dados["x1"].append(x1_val)
        dados["x2"].append(x2_val)
        dados["y"].append(y_ruido)

df = pd.DataFrame(dados)

# 2. Definindo a malha densa (grid) de points para estimar a superfície contínua
eixo_x1 = [i / 2 for i in range(0, int(n_lado * 2))]
eixo_x2 = [i / 2 for i in range(0, int(n_lado * 2))]

# Gera a lista de tuplas coordenadas [(x1_0, x2_0), (x1_0, x2_1), ...]
points_grid_2d = [(p1, p2) for p1 in eixo_x1 for p2 in eixo_x2]

# 3. Executando a suavização recursiva genérica passando a lista de X
resp = kernel_smoothing(
    df, 
    points=points_grid_2d, 
    x=["x1", "x2"],   # Passando os dois X como lista de strings
    y="y", 
    std=2.5, 
    iterations=8
)

# 4. Preparando os dados para plotagem 3D (Remapeando listas para matrizes)
len_x1, len_x2 = len(eixo_x1), len(eixo_x2)
X1_mesh = resp["x1"].values.reshape(len_x1, len_x2)
X2_mesh = resp["x2"].values.reshape(len_x1, len_x2)
Y_mesh = resp["y"].values.reshape(len_x1, len_x2)

# 5. Renderizando o gráfico tridimensional
fig = plt.figure(figsize=(15, 6))
ax = fig.add_subplot(111, projection='3d')

# Plota os points reais amostrados com ruído
ax.scatter(df["x1"], df["x2"], df["y"], color="black", alpha=0.4, s=15, label="Noisy sample")

# Plota a superfície estimada gerada pela regressão por kernel
surface = ax.plot_surface(
    X1_mesh, X2_mesh, Y_mesh, 
    cmap="coolwarm", 
    alpha=0.7, 
    edgecolor="none",
    zorder=5
)

# Customizações do cenário tridimensional
ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("Y")
fig.colorbar(surface, ax=ax, shrink=0.5, aspect=10, label="Estimated Value of Y")

ax.view_init(elev=30, azim=-45)  # Ajusta o ângulo de visão tridimensional
plt.legend()
plt.show()
```
